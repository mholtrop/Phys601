# Spiff shoots a pellet


In this imagined problem, Spaceman Spiff (yes, from Calvin and Hobbes) shoots a pellet from a railgun at 0.6$c$ to a target at the front of his spaceship, which is traveling at 0.5$c$ relative to Zephyr, who is observing the target practice.

What do the world lines for Spiff, Zephyr and the pellet look like on a Minkowsky space-time diagram? 

First we look at the space-time diagram for the single actor "Spiff". From Spiff's perspective, everything looks pretty normal, you see a nice t versus x plot. The slider at the bottom of the plot allows you to boost to a different reference frame. If we boost to -0.2$c$ (so boost to the left), then Spiff is following a slanted line of 0.2$c$ to the right, as it should be. 

Note however the red coordinates, which are still those from Spiff's perspective, and we see that they are now skewed in a very particular way. For our "observer" axis (the normal y-axis) the seconds are still seconds, but we now "see" the seconds for Spiff taking longer. This is just the time dilation. The skewed time axis for Spiff is such that from his perspective _our_ time is running slow, which is still time dilation.

## Spaceman Spiff shoots the pellet

Finally, we make a space-time diagram for the problem at hand. We initially take the point of view of Spiff, so Spiff is not moving (vertical line). Zephyr is moving to the left at 0.5$c$ (so negative). The railgun pellet is fired right when Spiff crosses Zephyr, at a velocity of 0.6$c$. You can now move the slider to Zephyr's reference frame (so their line is vertical) and see that the pellet still does not go faster than $c$. From the slope of the pellet you find the
velocity of the pellet in Zephyr's rest frame.

```python
plot = MinkowsiPlot()
plot.ShowAltAxes=False

spiff = Actor("Spiff",0.,0.,'red')
zephyr = Actor("Zephyr",-0.5,0.,'green')
pellet = Actor("pellet",0.6,0.,'blue')
pellet.PositiveOnly=True

plot.add_actor(spiff)
plot.add_actor(zephyr)
plot.add_actor(pellet)

fig = plot.make_figure()

fig.show()
```

Check that the math actually works!

To _compute_ the velocity of the pellet as Zephyr sees it, we use the formula for adding velocities:
$$ v_x' = \frac{v_x - u}{1 - u v_x/c^2}$$
where we have to be careful with the sign of $u$. In this case, the velocities _add_ so $u$ as written in the formula must be _negative_ as it is for the $u$ in the diagram. For the numbers given in the problem we then have
for the velocity of the pellet $v_x = 0.6c$ and the boost to Zephyr's reference frame $u= -0.5c$, which gives: 
$$ v_x' = \frac{0.6c+0.5c}{1 + (0.6c)(0.5c)/c^2} = 0.846c$$
Comparing this with the slope (hover over the last point of the pellet to get the numbers for $(x,t)$, we find:
$$ v_x' = \frac{7.9386}{9.3819} = 0.846c$$
The two agree (which is a bit of a relieve given the amount of code it took to make the diagram.)
