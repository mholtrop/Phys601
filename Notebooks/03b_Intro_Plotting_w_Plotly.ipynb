# Intro to Plotting with Plotly


Plotly is a relatively newer plotting package, though it is now already up to version 4. It is a commercial package turned open-source that is free to use. What is special about this package is that it uses your browser for the output. The package converts the Python into JavaScript, which then runs in your browser. This allow you to interact with the resulting graphs. If you pay them, they can also host your plots for you. In addition to Plotly, the same company has a product called Dash, which allows you to make the web pages truly interactive. Though that setup is not trivial, it is a great choise to display graphs that were created from live data from, for instance, a Raspberry Pi, using their Dash App. [Documentation: Plotly](https://plot.ly/python/) [Plotly main site](https://plot.ly)

## Getting started with Plotly

You can find more details for Plotly in the [Documentation](https://https://plot.ly/python). This package has undergone very rapid development and now has a huge number of features and plotting style. It is hard enough to keep track of everything that the Plot.ly people have created a handy, printable, [getting started cheat sheet](https://images.plot.ly/plotly-documentation/images/python_cheat_sheet.pdf)

To get started with Plot.ly, you need to import the package into Python. The part that you typically interact with is the sub-package, matplotlib.pyplot. A standard way to import the package is with
`import plotly.plotly as py`

```python
import plotly.graph_objects as go
import plotly.io as pio
import numpy as np
pio.renderers.default = "notebook_connected"
# pio.renderers.default = 'jupyterlab'
# This makes sure that the graphs will show in the notebook
# Alternate: 'browser' opens new browser window with the plots.
# For Jupyter-lab the uncommented second choice works better.
```

```python
x = np.linspace(-5,5,100)  # We want a really smooth plot with a 100 points instead of 15.
y = x*x                    # This computes an array y with each element filled with x-square.\
fig = go.Figure(data=go.Scatter(x=x, y=y))
fig.update_layout(
    title=go.layout.Title(
        text="Quadratic Curve",
        yanchor="top",
        y=0.86,
        xanchor="center",
        x=0.5,
        font=dict(size=24)),
    xaxis=go.layout.XAxis(
        title=go.layout.xaxis.Title(text="x Axis"),
        tickfont=dict(size=18),
    ),
    yaxis=go.layout.YAxis(
        title=go.layout.yaxis.Title(
            text="y Axis"),
        tickfont=dict(size=18),
    )
)
fig.show()
```

## Multiple lines
You can add multiple lines, with labels, by simply adding multiple plots in a single figure.

```python
x = np.linspace(-1,1,100)
y1 = x
y2 = x*x
y3 = x*x*x
y4 = x*x*x*x
# Note that the name, hovertext, and line modifiers are optional
fig = go.Figure(data=[
    go.Scatter(x=x, y=y1,name=r"$\text{Linear:}\ x$",hovertext="x",line=dict(color="red")),
    go.Scatter(x=x, y=y2,name=r"$\text{Quadratic:}\ x^2$",hovertext="x*x",line=dict(color="blue",dash="dash")),
    go.Scatter(x=x, y=y3,name=r"$\text{Qubic:}\ x^3$",hovertext="x*x*x",line=dict(color="green",dash="dashdot")),
    go.Scatter(x=x, y=y4,name=r"$\text{Forth power:}\ x^4$",hovertext="x*x*x*x")
    ])
fig.update_layout(
    title=go.layout.Title(
        text="Plots of Polynomials",
        yanchor="top",
        y=0.86,
        xanchor="center",
        x=0.5,
        font=dict(size=24)),
    xaxis=go.layout.XAxis(
        title=go.layout.xaxis.Title(text="x Axis", font=dict(size=20)),
        tickfont=dict(size=18),
    ),
    yaxis=go.layout.YAxis(
        title=go.layout.yaxis.Title(text="y Axis", font=dict(size=20)),
        tickfont=dict(size=18),
    )
)
fig.show()

x = np.linspace(-1,10,100)
y1 = x
y2 = x*x
y3 = x*x*x
y4 = x*x*x*x
# Note that the name, hovertext, and line modifiers are optional

# Note that the name, hovertext, and line modifiers are optional
fig = go.Figure(data=[
    go.Scatter(x=x, y=y1,name=r"$\text{Linear:}\ x$",hovertext="x",line=dict(color="red")),
    go.Scatter(x=x, y=y2,name=r"$\text{Quadratic:}\ x^2$",hovertext="x*x",line=dict(color="blue",dash="dash")),
    go.Scatter(x=x, y=y3,name=r"$\text{Qubic:}\ x^3$",hovertext="x*x*x",line=dict(color="green",dash="dashdot")),
    go.Scatter(x=x, y=y4,name=r"$\text{Forth power:}\ x^4$",hovertext="x*x*x*x")
    ])

fig.update_layout(
    title=go.layout.Title(
        text="Plots of Polynomials",
        yanchor="top",
        y=0.86,
        xanchor="center",
        x=0.5,
        font=dict(size=24)),
    xaxis=go.layout.XAxis(
        title=go.layout.xaxis.Title(text="x Axis", font=dict(size=20)),
        tickfont=dict(size=18),
    ),
    yaxis=go.layout.YAxis(
        title=go.layout.yaxis.Title(text="y Axis", font=dict(size=20)),
        tickfont=dict(size=18),
    ),
    xaxis_type = "log",
    yaxis_type = "log"
)
fig.show()
```

### Styling the graph
The SI notation style is *not* standard in Physics, so we should change that. It takes some searching on how to do this, but then in the end, the solution is fairly simple. Here is the last plot again.

```python
fig.update_layout(
    xaxis=go.layout.XAxis(
    title="x",
    type="log",
    exponentformat = "power"
    ),
    yaxis=go.layout.YAxis(
    title="f(x)",
    type="log",
    exponentformat = "power" # List of possible formats: "none" "e" "E" "power" "B" "SI"
    ))
fig.show()
```

## Parametrized plots
With most plotting packages, you can simply make a parametrized plot by just plotting the $x$ and $y$ coordinates.
A nice feature of Plot.ly is that you can add a "hovertext" to label each point, in this case with the number of your parameter. 

```python
t=np.linspace(-4,4,2**8+1)
x=np.real(np.power((-2.+0.j),t))
y=np.imag(np.power((-2.+0.j),t))
```

```python
print("The hovertext sets the text that will appear when you hover your cursor over a point.")
print("This is a nice way to make a parametrized plot, where the paramets shows up in the hover.")
fig = go.Figure(data=go.Scatter(x=x, y=y,hovertext=["t={:7.3f}".format(x) for x in t]))
fig.update_layout(
    title=go.layout.Title(
        text="Powers of -2.",
        yanchor="top",
        y=0.86,
        xanchor="center",
        x=0.5,
        font=dict(size=24),
    ),
    xaxis=go.layout.XAxis(
        title=go.layout.xaxis.Title(text="x Axis", font=dict(size=20)),
        tickfont=dict(size=18),
    ),
    yaxis=go.layout.YAxis(
        title=go.layout.yaxis.Title(text="Imaginary", font=dict(size=20)),
        tickfont=dict(size=18),
    )
)

fig.show()
```

```python
fig = go.Figure(data=go.Scatter3d(x=x, y=y,z=t,
                                  mode="lines"
                                  ))
fig.update_layout(
    title=go.layout.Title(
        text="Powers of -2.",
        yanchor="top",
        y=0.86,
        xanchor="center",
        x=0.5,
        font=dict(size=24)),
    xaxis=go.layout.XAxis(
        title=go.layout.xaxis.Title(text="Real", font=dict(size=20)),
        tickfont=dict(size=18),
    ),
    yaxis=go.layout.YAxis(
        title=go.layout.yaxis.Title(text="Imaginary", font=dict(size=20)),
        tickfont=dict(size=18),
    )
)
fig.show()
```
